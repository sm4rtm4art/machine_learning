# Encoder Feature Extraction: Beyond Text Generation

**Status**: 🚧 DRAFT - Concept documented, implementation pending

## The Core Idea

Most TrOCR tutorials do this:
```python
output = model.generate(pixel_values)  # Black box → text
```

We want to do this:
```python
embedding = model.encoder(pixel_values)  # Extract visual representation
# Now we have a 768-dim vector we can analyze, visualize, classify
```

## Why This Matters

1. **Interpretability** - See what the model "sees" before text generation
2. **Classification** - Route documents without running expensive decoder
3. **Debugging** - Visualize embedding space (t-SNE, UMAP)
4. **Hybrid Pipelines** - Feed embeddings to classical ML (SVM, see notebook 03)

## Key Code Snippet (Preview)

```python
from transformers import VisionEncoderDecoderModel, TrOCRProcessor

model = VisionEncoderDecoderModel.from_pretrained("microsoft/trocr-base-printed")
processor = TrOCRProcessor.from_pretrained("microsoft/trocr-base-printed")

def extract_embedding(image):
    """Extract visual embedding from encoder (skip decoder)."""
    pixel_values = processor(image, return_tensors="pt").pixel_values
    
    with torch.no_grad():
        encoder_outputs = model.encoder(pixel_values)
        # Pool sequence to single vector (mean pooling)
        embedding = encoder_outputs.last_hidden_state.mean(dim=1)
    
    return embedding.squeeze().numpy()  # Shape: (768,)
```
